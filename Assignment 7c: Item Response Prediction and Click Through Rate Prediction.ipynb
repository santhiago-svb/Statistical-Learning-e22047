1.1 The Two-Parameter Logistic (2PL) IRT model calculates success probability using:
  $$p_i(\theta) = \frac{1}{1 + e^{-a_i(\theta - b_i)}}$$Key Roles of Difficulty ($b_i$)Horizontal Shift: Moving $b_i$ translates the curve along the ability axis.  Higher $b_i$ (Right): Needs higher ability $\theta$ to answer correctly.  Lower $b_i$ (Left): Easier item; lower ability is sufficient.  50% Success Threshold: $b_i$ is the exact ability level where probability is 50%:
  $$p_i(b_i) = \frac{1}{1 + e^0} = 0.5$$Relative Ability Difference:$\theta > b_i \implies p_i(\theta) > 0.5$  $\theta < b_i \implies p_i(\theta) < 0.5$  Slope Anchor: $b_i$ anchors the center of the curve, where discrimination $a_i$ determines steepness (max slope $= \frac{a_i}{4}$).  




1.2 Key Components of Sequential LikelihoodSingle Response Likelihood ($y_k$): For an item with parameters $a_k$ and $b_k$, the likelihood of a single response $y_k \in \{0, 1\}$ is:
  $$L(y_k \mid \theta) = [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k}$$
where $p_k(\theta) = \frac{1}{1 + e^{-a_k(\theta - b_k)}}$.  Joint Likelihood Function ($\mathbf{y}^{(k)}$): Assuming local independence between item responses given $\theta$, the total likelihood after $k$ items is the product of individual response likelihoods:
  $$L(\mathbf{y}^{(k)} \mid \theta) = \prod_{i=1}^{k} [p_i(\theta)]^{y_i} [1 - p_i(\theta)]^{1 - y_i}$$Sequential Multiplication: The likelihood updates incrementally with each new item $k$ by scaling the previous joint likelihood:
  $$L(\mathbf{y}^{(k)} \mid \theta) = L(y_k \mid \theta) \cdot L(\mathbf{y}^{(k-1)} \mid \theta)$$

1.3 Key Components of the Update FormulaProportional Posterior Update: The updated posterior density after response $y_k$ is proportional to the current likelihood times the prior posterior density:
  $$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto L(y_k \mid \theta) \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$Explicit Likelihood Expansion: Substituting the single-item 2PL likelihood yields:
  $$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k} \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$Normalizing Constant ($Z_k$): To ensure a valid probability distribution, the posterior is normalized by integrating across all $\theta$:
  $$Z_k = \int_{-\infty}^{\infty} L(y_k \mid \theta) \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)}) \, d\theta$$$$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) = \frac{1}{Z_k} L(y_k \mid \theta) \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$

1.4 Key Mechanics of the Dynamic ShiftAsymmetric Likelihood Weighting: The likelihood function $L(y_k = 1 \mid \theta) = p_k(\theta)$ stays near $0$ for low $\theta$ and rises toward $1$ near $b_k$.  Suppression of Low Abilities: Multiplying the previous posterior $f_{\Theta \mid \mathbf{Y}^{(k-1)}}$ by this likelihood suppresses lower $\theta$ values while amplifying higher ones.  Rightward Posterior Shift: This multiplication pushes the mode (peak) of the updated posterior distribution sharply to the right.  Belief Reinforcement: The update aggressively boosts the system's estimate of the user's latent ability ($\theta$), reflecting strong evidence of high competence.  

1.5 Mathematical Role of $a_k$ in the Item Characteristic Curve (ICC)The discrimination parameter $a_k > 0$ determines the steepness of the logistic probability curve:  $$p_k(\theta) = \frac{1}{1 + e^{-a_k(\theta - b_k)}}$$The maximum slope of $p_k(\theta)$ occurs precisely at $\theta = b_k$ and is proportional to $a_k$:$$\left. \frac{d}{d\theta} p_k(\theta) \right\vert{}_{\theta = b_k} = \frac{a_k}{4}$$As $a_k \to \infty$: The logistic curve approaches a step function (Heaviside step function) centered at $b_k$.As $a_k \to 0$: The logistic curve becomes flat at $p_k(\theta) \approx 0.5$ for all $\theta$, carrying no discriminative power.2. Impact on the Posterior Update and Variance ReductionThe Bayesian posterior update multiplies the prior density by the likelihood term $L(y_k \mid \theta) = [p_k(\theta)]^{y_k} [1 - p_k(\theta)]^{1 - y_k}$:  $$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto L(y_k \mid \theta) \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$The sharp gradient introduced by $a_k$ directly influences the shape and width of the updated distribution:A. High Discrimination (Large $a_k$)Sharp Division of Likelihood:If $y_k = 1$, the likelihood $p_k(\theta)$ drops rapidly toward $0$ for all $\theta < b_k$ and climbs sharply to $1$ for $\theta > b_k$.If $y_k = 0$, $1 - p_k(\theta)$ drops rapidly toward $0$ for all $\theta > b_k$.Variance Reduction:
Multiplying the prior density by this steep likelihood drastically zeroes out a large portion of the parameter space. This results in a sharp, tall posterior peak with a significantly smaller variance, indicating high confidence/certainty in the estimate.  B. Low Discrimination (Small $a_k$)Flat Likelihood Gradient:The likelihood curve changes very slowly across the $\theta$ domain.Minimal Information Gain:
Multiplying the prior by a nearly flat likelihood scales the distribution almost uniformly. The updated posterior retains roughly the same width and variance as the prior $f_{\Theta \mid \mathbf{Y}^{(k-1)}}$, reflecting very little change in system certainty.  3. Quantification via Item Information Functions (IIF)In psychometrics, the information $I_k(\theta)$ provided by an item is defined by its Fisher Information:  $$I_k(\theta) = a_k^2 \cdot p_k(\theta) \cdot [1 - p_k(\theta)]$$Quadratic Dependence on $a_k$: The information provided by an item scales quadratically with $a_k$ ($I_k \propto a_k^2$). Doubling $a_k$ quadruples the maximum information yielded at $\theta = b_k$.Asymptotic Variance: By the Bernstein–von Mises theorem, as item responses accumulate, the posterior variance is inversely proportional to the total accumulated information:$$\text{Var}(\Theta \mid \mathbf{y}^{(k)}) \approx \frac{1}{\sum_{i=1}^k I_i(\hat{\theta})}$$Summary: High discrimination items ($a_k \gg 1$) inject large amounts of Fisher information into the update step, causing a steep drop in posterior variance (narrowing the probability curve), whereas low discrimination items ($a_k \to 0$) contribute minimal information and leave posterior certainty virtually unchanged.  

```python
import numpy as np
import scipy.stats as stats
import plotly.graph_objects as go

np.random.seed(42)

theta_true = 0.35
n_impressions = 100
steps = list(range(n_impressions + 1))

alpha_param = 1
beta_param = 1

running_bayes = [alpha_param / (alpha_param + beta_param)]
running_map = [0.0]

for k in range(1, n_impressions + 1):
    y_k = 1 if np.random.uniform(0, 1) < theta_true else 0

    alpha_param += y_k
    beta_param += (1 - y_k)

    theta_bayes_k = alpha_param / (alpha_param + beta_param)
    theta_map_k = (alpha_param - 1) / (alpha_param + beta_param - 2) if (alpha_param > 1 and beta_param > 1) else 0.0

    running_bayes.append(theta_bayes_k)
    running_map.append(theta_map_k)

fig = go.Figure()
fig.add_hline(y=theta_true, line_dash="dash", line_color="red", line_width=2,
              annotation_text=f"True CTR (\u03b8={theta_true})")

fig.add_trace(go.Scatter(x=steps, y=running_bayes, mode='lines', name='Exact Posterior Mean'))
fig.add_trace(go.Scatter(x=steps, y=running_map, mode='lines', name='Exact MAP Estimate', line=dict(dash='dot')))

fig.update_layout(title="Analytical Beta-Binomial Conjugate Update Timeline",
                  xaxis_title="Number of User Impressions (k)", yaxis_title="Estimated Conversion Rate (\u03b8)",
                  template="plotly_white")
fig.show()
```

1.7 Key Convergence MechanicsSequential Variance Reduction: As the user answers more items ($k \to 20$), each response multiplies the posterior density by an additional likelihood term, progressively narrowing the posterior distribution.  Estimator Alignment: Initial estimates are heavily influenced by the prior distribution ($\mathcal{N}(0, 1)$). However, as data accumulates, both $\hat{\theta}_{\text{Bayes}}^{(k)}$ and $\hat{\theta}_{\text{MAP}}^{(k)}$ converge toward the true parameter value ($\theta_{\text{true}} = 0.75$).  Impact of Item Discrimination: Items with higher $a_k$ parameters generate sharper likelihood updates, causing larger step-wise corrections and accelerating convergence toward $\theta_{\text{true}}$.  Measurement Precision: The shrinking distance between the point estimates and $\theta_{\text{true}}$ demonstrates that the system's belief stabilizes and measurement error diminishes with increasing sample size.  

2.1The Beta distribution $\text{Beta}(\alpha, \beta)$ serves as a flexible prior over the continuous interval $\theta \in [0, 1]$.  Key Shapes & Prior InterpretationsUninformative Prior $(\alpha=1, \beta=1)$:Equivalent to a Uniform distribution $U(0, 1)$.  All click-through rates $\theta$ are equally likely; represents complete prior uncertainty.  Right-Skewed Prior $(\alpha=2, \beta=8)$:Mode/mass is shifted toward $0$ (since $\beta > \alpha$).  Represents a cautious belief that CTRs are naturally low.  Left-Skewed Prior $(\alpha=8, \beta=2)$:Mode/mass is shifted toward $1$ (since $\alpha > \beta$).  Represents an optimistic belief in a high click-through rate.  

2.2In CTR tracking, $\theta \in [0, 1]$ represents the probability that an impression results in a click.  Key Components of Sequential LikelihoodSingle Impression Likelihood ($y_k$): Modeled as a Bernoulli trial for response $y_k \in \{0, 1\}$:
  $$L(y_k \mid \theta) = \theta^{y_k} (1 - \theta)^{1 - y_k}$$Click ($y_k = 1$): $L(y_k \mid \theta) = \theta$Non-Click ($y_k = 0$): $L(y_k \mid \theta) = 1 - \theta$Joint History Likelihood ($\mathbf{y}^{(k)}$): Assuming independent impressions given $\theta$, the joint probability across history $\mathbf{y}^{(k)} = (y_1, \dots, y_k)$ is the product of individual outcomes:
  $$L(\mathbf{y}^{(k)} \mid \theta) = \prod_{i=1}^{k} \theta^{y_i} (1 - \theta)^{1 - y_i} = \theta^{\sum_{i=1}^k y_i} (1 - \theta)^{\sum_{i=1}^k (1 - y_i)}$$Sufficient Statistic ($C_k$): Letting $C_k = \sum_{i=1}^k y_i$ (total clicks) simplifies the joint likelihood to:
  $$L(\mathbf{y}^{(k)} \mid \theta) = \theta^{C_k} (1 - \theta)^{k - C_k}$$
Total clicks $C_k$ and total trials $k$ store all necessary information to estimate $\theta$ without keeping individual response histories.  Recursive Updating: The likelihood updates continuously with each new impression:
  $$L(\mathbf{y}^{(k)} \mid \theta) = L(y_k \mid \theta) \cdot L(\mathbf{y}^{(k-1)} \mid \theta)$$

2.3Key Components of Closed-Form UpdatesAnalytical Proof of Conjugacy:Prior at step $k-1$: $f(\theta \mid \mathbf{y}^{(k-1)}) \propto \theta^{\alpha_{k-1} - 1} (1 - \theta)^{\beta_{k-1} - 1}$.  Single-step Bernoulli likelihood: $L(y_k \mid \theta) = \theta^{y_k} (1 - \theta)^{1 - y_k}$.  Posterior kernel update:
  $$f(\theta \mid \mathbf{y}^{(k)}) \propto \left[\theta^{y_k} (1 - \theta)^{1 - y_k}\right] \cdot \left[\theta^{\alpha_{k-1} - 1} (1 - \theta)^{\beta_{k-1} - 1}\right] = \theta^{(\alpha_{k-1} + y_k) - 1} (1 - \theta)^{(\beta_{k-1} + 1 - y_k) - 1}$$This matches a $\text{Beta}(\alpha_k, \beta_k)$ density, proving Beta-Binomial conjugacy.  Exact Parameter Update Rules:$$\alpha_k = \alpha_{k-1} + y_k, \qquad \beta_k = \beta_{k-1} + (1 - y_k)$$Posterior Mean Formula: The expected CTR after $k$ impressions is given analytically by:
  $$\mathbb{E}[\Theta \mid \mathbf{y}^{(k)}] = \frac{\alpha_k}{\alpha_k + \beta_k} = \frac{\alpha_0 + C_k}{\alpha_0 + \beta_0 + k}$$

2.4In Beta-Binomial CTR modeling, each user interaction ($y_k \in \{0, 1\}$) dynamically adjusts the posterior density over $\theta$.  Key Mechanics of Dynamic ShiftingClick ($y_k = 1$): Increments $\alpha_k$ by $1$ ($\alpha_k = \alpha_{k-1} + 1$), shifting the distribution mode to the right (higher estimated CTR).  Non-Click ($y_k = 0$): Increments $\beta_k$ by $1$ ($\beta_k = \beta_{k-1} + 1$), shifting the distribution mode to the left (lower estimated CTR).  Variance Reduction: Every impression increases the total sum $(\alpha_k + \beta_k)$ by $1$, making the posterior curve narrower and taller as uncertainty shrinks over time.  Computational Efficiency ($O(1)$ vs. IRT):Beta-Binomial conjugate updates run in $O(1)$ constant time using simple scalar additions.  Unlike non-conjugate 2PL IRT models, conjugate updates require no numerical integration or grid evaluations, making them ideal for real-time tracking.  

2.5Key Point EstimatorsRunning Posterior Mean ($\hat{\theta}_{\text{Bayes}}^{(k)}$): The expected value of the Beta posterior distribution:
  $$\hat{\theta}_{\text{Bayes}}^{(k)} = \mathbb{E}[\Theta \mid \mathbf{y}^{(k)}] = \frac{\alpha_k}{\alpha_k + \beta_k}$$Running Maximum A Posteriori ($\hat{\theta}_{\text{MAP}}^{(k)}$): The mode of the Beta posterior distribution (for $\alpha_k, \beta_k > 1$):
  $$\hat{\theta}_{\text{MAP}}^{(k)} = \text{Mode}(\Theta \mid \mathbf{y}^{(k)}) = \frac{\alpha_k - 1}{\alpha_k + \beta_k - 2}$$Analytical Simplicity: Both estimates depend directly on the updated parameter counts ($\alpha_k, \beta_k$), allowing instant $O(1)$ computation without numerical integration.  

2.6 Key Convergence MechanicsPrior Washing Effect: As $k \to \infty$, the sample click ratio $\frac{C_k}{k}$ dominates the posterior equation $\hat{\theta}_{\text{Bayes}}^{(k)} = \frac{\alpha_0 + C_k}{\alpha_0 + \beta_0 + k}$, rendering the initial prior $(\alpha_0, \beta_0)$ negligible.Estimator Alignment: The gap between the mean ($\hat{\theta}_{\text{Bayes}}^{(k)}$) and MAP ($\hat{\theta}_{\text{MAP}}^{(k)}$) shrinks at a rate of $O(1/k)$, causing both estimators to collapse onto each other and converge strictly to $\theta_{\text{true}}$.Variance Reduction: Posterior variance contracts at a rate of $\mathcal{O}(1/k)$, squeezing the credible interval tightly around $\theta_{\text{true}}$ as measurement precision stabilizes.
