```python
from data_analysis import DataInspector, PlottingMethods
import pandas as pd
```

```python
# Create an instance of the DataInspector class
inspector = DataInspector()
```

```python
# Load the classic Titanic dataset directly from a public URL
raw_titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
demo_df = pd.read_csv(raw_titanic_url)
```

```python
# Inject the dataframe into the pipeline engine
inspector.load_dataframe(demo_df)

inspector.display_summary()
```

```python
# Impute age using the median and categorical Embarked targets with mode metrics
inspector.handle_missing_values(columns=['Age'], strategy='median')
inspector.handle_missing_values(columns=['Embarked'], strategy='mode')
```

```python
# Automatically drop extreme outlier variance along the Fare coordinate system
inspector.handle_outliers(columns=['Fare'], action='delete')
```

```python
# Extract Z-score scaled numeric targets alongside one-hot structural flags
unified_df = inspector.generate_unified_dataset(
    numeric_cols=['Age', 'Fare'],
    categorical_cols=['Sex', 'Embarked'],
    num_strategy='standard',
    cat_strategy='onehot'
)
print("\n--- Model-Ready Unified Data Output Vector Shape ---")
print(unified_df.shape)
display(unified_df.head(5))
```

```python
# 3. View structural frequency weighting splits
inspector.plot_categorical_frequency('Embarked')
```
