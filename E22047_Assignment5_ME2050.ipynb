## 1. Bias Quantification in the Maximum Likelihood Covariance Estimator

We begin with the Maximum Likelihood Estimator (MLE) for the population covariance matrix $\boldsymbol{\Sigma}$, defined as:

$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

where the empirical sample mean is denoted by $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$.

To evaluate the bias properties of this estimator, we express the deviations relative to the unobserved, true population mean $\boldsymbol{\mu}$:

$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this decomposition back into the summation yields:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the outer product term-by-term, we obtain:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

By recognizing that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$, the central cross-product terms simplify, leading to:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Applying the expectation operator $\mathbb{E}[\cdot]$ across the equation gives:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

Using the definitions of the population statistical moments:
* $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these identities back into our expectation formula:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

Since $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the MLE technique introduces a systematic downward bias. This underestimation occurs because minimizing deviations relative to the *computed sample mean* rather than the *true population mean* artificially restricts the perceived variance of the dataset.

### Correcting Bias via Bessel's Correction

To construct an unbiased covariance estimator ($\mathbf{S}$), we scale the MLE matrix by the inversion of the bias factor, namely $\frac{n}{n-1}$:

$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Verifying the expected value confirms the elimination of the systemic bias:

$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

## 2. Diagnostics and Operational Error Classification

* **Type I Error ($\alpha$):** A **False Positive**. This occurs when the structural health monitoring system flags an anomaly or structural failure (rejects $H_0$), even though the physical system is operating normally and remains undamaged.
* **Type II Error ($\beta$):** A **False Negative**. This occurs when a structural defect or physical crack exists (rejects $H_1$), but the diagnostic routine fails to trigger an alarm (fails to reject $H_0$).

### Operational Implications of an Ultra-Conservative Threshold ($\alpha = 0.0001$)

Setting an exceptionally strict significance level to avoid false alarms shifts the decision boundary far out into the tails of the null distribution. This creates two primary cascading effects:

1. **Degradation of Statistical Power ($1-\beta$):** Because the threshold for detecting anomalies is heavily constrained, emerging structural changes or minor flaws will go unnoticed. Consequently, the probability of a missed detection ($\beta$) surges, leading to a severe **collapse in the statistical power ($1-\beta$) of the test**.
2. **Expansion of Confidence Ellipsoids:** The corresponding multi-dimensional confidence envelope broadens to a level of $1-\alpha = 0.9999$. Structurally, the **geometric volume of the healthy baseline ellipsoid swells substantially**. Because this boundary expands to encompass nearly all data variations, the diagnostic system becomes highly desensitized to actual structural degradation.

## Slutsky's Theorem and Asymptotic Covariance Substitution

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables (or vectors). If

$$
X_n \xrightarrow{d} X
$$

and

$$
Y_n \xrightarrow{p} c,
$$

where $c$ is a constant scalar or matrix, then Slutsky's Theorem states:

$$
X_n + Y_n \xrightarrow{d} X + c,
$$

$$
X_n Y_n \xrightarrow{d} cX,
$$

$$
\frac{X_n}{Y_n} \xrightarrow{d} \frac{X}{c},
\qquad c \neq 0.
$$

The Lindeberg--Lévy Multivariate Central Limit Theorem asserts that

$$

$$

$$
\mathbf{S}

\boldsymbol{\Sigma}
$$

which implies, by continuity of the matrix inverse square root mapping,

$$
\mathbf{S}^{-1/2}
\xrightarrow{p}
\boldsymbol{\Sigma}^{-1/2}.
$$

Applying Slutsky's Theorem,

$$
\mathbf{S}^{-1/2}
\left[
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\right]
\xrightarrow{d}
\boldsymbol{\Sigma}^{-1/2}

\left(
\mathbf{0},
\boldsymbol{\Sigma}
\right).
$$

Since

$$
\boldsymbol{\Sigma}^{-1/2}
\,\mathcal{N}
\left(
\mathbf{0},
\boldsymbol{\Sigma}
\right)
=
\mathcal{N}
\left(
\mathbf{0},
\mathbf{I}
\right),
$$

it follows that

$$
\mathbf{S}^{-1/2}
\left[
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\right]
\xrightarrow{d}
\mathcal{N}
\left(
\mathbf{0},
\mathbf{I}
\right).
$$

Therefore, replacing the unknown population covariance matrix $\boldsymbol{\Sigma}$ with its consistent estimator $\mathbf{S}$ does not alter the asymptotic distribution.

Consequently,

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\approx
\mathcal{N}
\left(
\mathbf{0},
\mathbf{S}
\right),
$$

which implies

$$
\widehat{\boldsymbol{\mu}}_n
\approx
\mathcal{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right).
$$

Thus, the empirical covariance matrix $\mathbf{S}$ may be substituted for $\boldsymbol{\Sigma}$ while preserving the asymptotic stability of the multivariate Gaussian diagnostic envelope.

#### 1. Hotelling's $T^2$ Statistic
The Mahalanobis distance in the reduced principal subspace for a given truncation boundary $k$ is defined as:

$$T^2 = \sum_{j=1}^{k} \frac{z_{j}^2}{\hat{\lambda}_j}$$

Because the scores $z_j$ are uncorrelated and have an empirical variance exactly equal to their corresponding eigenvalue $\hat{\lambda}_j$, the standardized score variable $v_j = \frac{z_j}{\sqrt{\hat{\lambda}_j}}$ has a variance of $1$.

The expectation of a sum of squared standard normal-like variables equals the degrees of freedom:

$$\mathbb{E}[T^2] = \sum_{j=1}^{k} \mathbb{E}\left[\frac{z_{j}^2}{\hat{\lambda}_j}\right] = \sum_{j=1}^{k} 1 = k$$

$$Q = \sum_{j=k+1}^{m} z_{j}^2$$

$$\mathbb{E}[Q] = \sum_{j=k+1}^{m} \mathbb{E}[z_{j}^2] = \sum_{j=k+1}^{m} \hat{\lambda}_j$$

$$\mathbb{E}[Q] = 4 - \sum_{j=1}^{k} \hat{\lambda}_j$$

## Interpretation of PCA and Factor Analysis Results

### 1. What does a Uniqueness value close to $100\%$ tell us about `Sensor\_4`?

In Factor Analysis, the variance of each sensor is decomposed into:

$$
\text{Total Variance}
=
\text{Common Variance}
+
\text{Unique Variance}.
$$

A uniqueness value

$$
\varphi^2 \approx 0.98
$$

means that approximately $98\%$ of the variance of `Sensor\_4` cannot be explained by the common latent factors that drive the rest of the system.

Therefore, the signal from `Sensor\_4` is largely independent of the shared process dynamics. Physically, this suggests that the sensor is dominated by local noise, measurement error, calibration drift, electronic interference, or other sensor-specific effects rather than meaningful plant-wide behavior.

Consequently, `Sensor\_4` contributes very little information about the underlying structure of the process.

---

### 2. How does the large localized noise of `Sensor\_4` inflate PCA eigenvalues?

PCA seeks directions that maximize total variance. Mathematically, PCA is based on the covariance matrix

$$
\mathbf{\Sigma},
$$

whose eigenvalues satisfy

$$
\mathbf{\Sigma}\mathbf{v}_i
=
\lambda_i \mathbf{v}_i.
$$

The eigenvalue

$$
\lambda_i
$$

represents the amount of variance captured by the corresponding principal component.

Since PCA does not distinguish between common process variation and sensor-specific noise, any variable with extremely large variance contributes strongly to the covariance matrix.

For `Sensor\_4`,

$$
\text{Variance}(\text{Sensor}_4)
=
\text{Signal}
+
\text{Noise},
$$

and because the noise component is dominant,

$$
\text{Variance}(\text{Sensor}_4)
\approx
\text{Noise Variance}.
$$

As a result, PCA interprets this noise as important variance and allocates part of the leading principal components to explain it. This artificially increases the largest eigenvalues and may distort the principal component directions.

In effect, PCA is "tricked" into treating random sensor noise as meaningful system structure.

---

### 3. Why is this dangerous for an anomaly detection system?

An anomaly detection framework built solely on PCA assumes that the largest principal components represent the most important process behavior.

However, if a high-variance noisy sensor dominates the covariance structure, then

$$
\lambda_1,\lambda_2,\ldots
$$

may partially represent sensor noise rather than true plant dynamics.

This creates several risks:

1. **False Alarms**

   Random fluctuations in `Sensor\_4` may be interpreted as significant process changes, causing unnecessary anomaly alerts.

2. **Missed Faults**

   Genuine process abnormalities occurring in lower-variance but physically important variables may be hidden because PCA allocates excessive attention to the noisy sensor.

3. **Misleading Process Monitoring**

   Engineers may incorrectly conclude that the dominant principal components describe real plant behavior when they are actually capturing sensor-specific noise.

4. **Reduced Model Reliability**

   The anomaly detection model becomes sensitive to sensor degradation, calibration errors, and measurement disturbances rather than actual operational faults.

---

### Conclusion

The PCA results suggest a highly structured system because the first two principal components explain a large percentage of total variance. However, the Factor Analysis results reveal that `Sensor\_4` possesses

$$
\varphi^2 > 98\%.
$$

This indicates that most of its variance is unique, sensor-specific noise rather than common process information.

Therefore, Factor Analysis provides an important warning that PCA alone may overestimate the importance of `Sensor\_4`. For reliable industrial anomaly detection, PCA should be supplemented with methods that separate common process variation from sensor-specific noise.

````python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================================
# COMBINED SUBSPACE ENGINES (PCA & FA PIPELINES)
# =====================================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                        titleside="top"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }

# =====================================================================
# DATA GENERATOR & PIPELINE EXECUTION
# =====================================================================
if __name__ == "__main__":
    # Set execution seed for perfect reproducible properties
    np.random.seed(42)
    n_samples = 2500

    # 1. Generate underlying, independent true physical driving forces (Latent Factors)
    f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
    f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

    # 2. Construct the Observable Multi-Sensor Array with targeted noise profiles
    # Sensors 1 and 2 couple strongly to the primary process (f1)
    s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
    s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)

    # Sensor 3 couples strongly to the secondary process (f2)
    s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

    # Sensor 4 suffers an extreme electrical/instrumentation fault (massive localized white noise)
    s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

    # 3. Compile into the final evaluation DataFrame
    df_asset = pd.DataFrame(
        data=np.vstack([s1, s2, s3, s4]).T,
        columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
    )
    # Instantiate Engine
    engine = DualSubspaceDiagnosticsEngine(df_asset)

    # Run PCA Engine (1x5 Dashboard)
    pca_metrics = engine.compute_empirical_pca(show_plot=True)

    # Run FA Engine (2x2 Dashboard, Retaining k=2 Common Physical Factors)
    fa_metrics = engine.compute_empirical_fa(k=2, show_plot=True)
````

### Solution for Question 1: The Total Variance Illusion

## Interpretation of PCA and Factor Analysis Results

### 1. What does a Uniqueness value close to $100\%$ tell us about `Sensor\_4`?

In Factor Analysis, the variance of each sensor is decomposed into:

$$
\text{Total Variance}
=
\text{Common Variance}
+
\text{Unique Variance}.
$$

A uniqueness value

$$
\varphi^2 \approx 0.98
$$

means that approximately $98\%$ of the variance of `Sensor\_4` cannot be explained by the common latent factors that drive the rest of the system.

Therefore, the signal from `Sensor\_4` is largely independent of the shared process dynamics. Physically, this suggests that the sensor is dominated by local noise, measurement error, calibration drift, electronic interference, or other sensor-specific effects rather than meaningful plant-wide behavior.

Consequently, `Sensor\_4` contributes very little information about the underlying structure of the process.

---

### 2. How does the large localized noise of `Sensor\_4` inflate PCA eigenvalues?

PCA seeks directions that maximize total variance. Mathematically, PCA is based on the covariance matrix

$$
\mathbf{\Sigma},
$$

whose eigenvalues satisfy

$$
\mathbf{\Sigma}\mathbf{v}_i
=
\lambda_i \mathbf{v}_i.
$$

The eigenvalue

$$
\lambda_i
$$

represents the amount of variance captured by the corresponding principal component.

Since PCA does not distinguish between common process variation and sensor-specific noise, any variable with extremely large variance contributes strongly to the covariance matrix.

For `Sensor\_4`,

$$
\text{Variance}(\text{Sensor}_4)
=
\text{Signal}
+
\text{Noise},
$$

and because the noise component is dominant,

$$
\text{Variance}(\text{Sensor}_4)
\approx
\text{Noise Variance}.
$$

As a result, PCA interprets this noise as important variance and allocates part of the leading principal components to explain it. This artificially increases the largest eigenvalues and may distort the principal component directions.

In effect, PCA is "tricked" into treating random sensor noise as meaningful system structure.

---

### 3. Why is this dangerous for an anomaly detection system?

An anomaly detection framework built solely on PCA assumes that the largest principal components represent the most important process behavior.

However, if a high-variance noisy sensor dominates the covariance structure, then

$$
\lambda_1,\lambda_2,\ldots
$$

may partially represent sensor noise rather than true plant dynamics.

This creates several risks:

1. **False Alarms**

   Random fluctuations in `Sensor\_4` may be interpreted as significant process changes, causing unnecessary anomaly alerts.

2. **Missed Faults**

   Genuine process abnormalities occurring in lower-variance but physically important variables may be hidden because PCA allocates excessive attention to the noisy sensor.

3. **Misleading Process Monitoring**

   Engineers may incorrectly conclude that the dominant principal components describe real plant behavior when they are actually capturing sensor-specific noise.

4. **Reduced Model Reliability**

   The anomaly detection model becomes sensitive to sensor degradation, calibration errors, and measurement disturbances rather than actual operational faults.

---

### Conclusion

The PCA results suggest a highly structured system because the first two principal components explain a large percentage of total variance. However, the Factor Analysis results reveal that `Sensor\_4` possesses

$$
\varphi^2 > 98\%.
$$

This indicates that most of its variance is unique, sensor-specific noise rather than common process information.

Therefore, Factor Analysis provides an important warning that PCA alone may overestimate the importance of `Sensor\_4`. For reliable industrial anomaly detection, PCA should be supplemented with methods that separate common process variation from sensor-specific noise.

### Solution for Question 2: Decoupling Structural Loading via Rotation

* **Part 1 (Mathematical Jumble vs. Simple Structure):** Because PCA is bound to strict, orthogonal variance maximization, look at what happens to the remaining signals on `PC 2` and `PC 4`. `PC 2` and `PC 4` exhibit high loading values across *both* `Sensor_1` and `Sensor_2` simultaneously, while also bleeding into `Sensor_3`. This cross-loading represents a mathematically forced mixture of signals. Factor Analysis applies a **Varimax rotation** to eliminate this. By maximizing loading variance within columns, it completely isolates the signals: `Sensor_1` and `Sensor_2` group cleanly into `Factor 1`, while `Sensor_3` shifts entirely into `Factor 2`.
* **Part 2 (Operational Interpretability):** The PCA loading matrix forces a complex interpretation. If a diagnostic alarm trips on `PC 2`, an engineer has to figure out a messy combination of Sensors 1, 2, and 3. With the rotated FA framework, the layout is intuitive. `Factor 1` represents the isolated physical process causing redundancy in `Sensor_1` and `Sensor_2`, while `Factor 2` maps directly to the independent process monitored by `Sensor_3`. Troubleshooting root causes becomes instantaneous.

### Solution for Question 4: Operational Trade-offs in System Health Monitoring

## Comparison of PCA and Factor Analysis for Predictive Maintenance

### 1. PCA Strategy: Monitoring the Principal Subspace

The PCA approach models the dominant variation of the system using a reduced set of principal components.

Two common monitoring statistics are:

#### Hotelling's \(T^2\)

Measures variation within the principal component subspace:

$$
T^2
=
\mathbf{t}^{T}
\mathbf{\Lambda}^{-1}
\mathbf{t},
$$

where

$$
\mathbf{t}
=
\text{principal component score vector},
$$

and

$$
\mathbf{\Lambda}
=
\text{diagonal matrix of retained eigenvalues}.
$$

#### Q Statistic (Squared Prediction Error)

Measures variation outside the retained PCA subspace:

$$
Q
=
\mathbf{e}^{T}\mathbf{e},
$$

where

$$
\mathbf{e}
=
\mathbf{x}
-
\hat{\mathbf{x}}
$$

is the reconstruction residual.

The combination of \(T^2\) and \(Q\) allows PCA to monitor both structured variation and unexplained residual behavior.

---

### 2. FA Strategy: Monitoring Latent Factor Scores

Factor Analysis assumes that the observed sensor measurements can be decomposed as

$$
\mathbf{x}
=
\mathbf{\Lambda}\mathbf{f}
+
\boldsymbol{\varepsilon},
$$

where

$$
\mathbf{f}
=
\text{latent factor vector},
$$

and

$$
\boldsymbol{\varepsilon}
=
\text{sensor-specific noise}.
$$

The variance decomposition is

$$
\text{Total Variance}
=
\text{Common Variance}
+
\text{Unique Variance}.
$$

The uniqueness of sensor \(i\) is

$$
\varphi_i^2.
$$

Monitoring is performed directly on the estimated factor scores

$$
\hat{\mathbf{f}},
$$

which represent the underlying process behavior after separating sensor-specific noise.

---

### 3. Response to a Sensor Calibration Failure

Suppose Sensor 4 loses calibration or experiences an electrical short.

Factor Analysis reports

$$
\varphi_4^2 > 98\%.
$$

This means that more than \(98\%\) of the variance of Sensor 4 is unique variance rather than common process variance.

Therefore,

$$
\text{Common Variance}
\ll
\text{Unique Variance}.
$$

The sensor contributes very little information about the true process state.

---

### Effect on PCA

PCA does not distinguish between common variance and unique variance.

A faulty sensor with very large variance contributes directly to the covariance matrix

$$
\mathbf{\Sigma}.
$$

Consequently,

$$
\mathbf{\Sigma}\mathbf{v}_i
=
\lambda_i \mathbf{v}_i
$$

may produce inflated eigenvalues because the sensor fault increases total variance.

As a result:

1. Principal component directions may rotate.
2. Hotelling's \(T^2\) may increase unnecessarily.
3. False alarms may occur.
4. Process interpretation may become distorted.

Thus PCA can be heavily influenced by a single malfunctioning sensor.

---

### Effect on Factor Analysis

Factor Analysis explicitly separates common variation from sensor-specific variation.

Because

$$
\varphi_4^2 > 98\%,
$$

the abnormal behavior of Sensor 4 is assigned primarily to

$$
\boldsymbol{\varepsilon},
$$

rather than to the latent factors.

Therefore the factor scores

$$
\hat{\mathbf{f}}
$$

remain relatively stable unless the underlying process itself changes.

This greatly reduces the influence of a faulty sensor on the monitoring system.

---

### Operational Comparison

| Feature | PCA Strategy | FA Strategy |
|----------|-------------|-------------|
| Uses total variance | Yes | No |
| Separates sensor noise from process behavior | No | Yes |
| Sensitive to faulty sensors | High | Low |
| Robust to calibration drift | Moderate | High |
| Robust to electrical shorts | Moderate | High |
| Interpretability of process drivers | Moderate | High |

---

### Conclusion

The Factor Analysis strategy is more robust against a single sensor losing calibration or experiencing an electrical short.

The key reason is the Sensor Uniqueness Noise Floor metric

$$
\varphi^2.
$$

Since Sensor 4 has

$$
\varphi_4^2 > 98\%,
$$

Factor Analysis recognizes that most of its variance is sensor-specific noise and prevents that variance from contaminating the latent process factors.

In contrast, PCA treats all variance equally and may incorrectly interpret the faulty sensor's noise as meaningful process variation. Consequently, the PCA monitoring framework is more vulnerable to false alarms, distorted principal components, and unreliable anomaly detection when a sensor fails.
